```python
# 1. model 설정
model = init_chat_model("openai:gpt-5.6-luna", reasoning_effort="none")
```

#### 시나리오 설정

- 질문에 대한 답변을 강원도 사투리로 변경한 후 결과는 .md으로 저장
1. 질문에 대한 답변을 한다
2. 답변을 강원도 사투리로 변환한다.
3. 결과를 마크다운으로 저장한다.



```python
@tool
def change_text(question:str)  :
    """ 
    사용자가 요청 시 답변을 받아서 강원도 사투리로 변환한다
    """

    result = "강원도 사투리" + question + "변환했어요"
    return result
# 강원도 사투리 사용
```

```python
# # 결과를 마크다운으로 저장

def save_md(text : str): # 답변을 받아서 text으로 받음
    """ 
    최종 결과를 받아서 마크다운 문서로 저장
    파일명은 result.md
    
    """
    with open("result.md", "w", encoding="utf-8-sig") as file:
        file.write(text)
    return "저장 완료: result.md"
```

```python
agent = create_agent(


    model=model,
    tools=[change_text, save_md],
    system_prompt = """ 너는 사용자가 질문을 하면 답변을 하고, 적절하게 도구를 호출하여 답변하는 비서야"""
    
)
```

```python
# agent 호출용 함수
def ask(question):
    response = agent.invoke({"messages" : [ HumanMessage(question) ]})
    return response
```

```python
result = ask("안녕하세요")
result['messages']
```

```python
result = ask("어떻게 할까?")
result['messages']
```
