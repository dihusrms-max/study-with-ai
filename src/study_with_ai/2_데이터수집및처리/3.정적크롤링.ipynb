### 정적 웹 크롤링

- 페이지 형태가 정적페이지 일 때 (추가 정보 요청 없을 시)

```python
from bs4 import BeautifulSoup
import requests
#url = file:///C:/study-with-ai/src/study_with_ai/2_%EB%8D%B0%EC%9D%B4%ED%84%B0%EC%88%98%EC%A7%91%EB%B0%8F%EC%B2%98%EB%A6%AC/static_crawling_practice.html

name ="조승우"
url = f"https://ko.wikipedia.org/wiki/{name}"

headers = {
      "User-Agent": (
          "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
          "AppleWebKit/537.36 (KHTML, like Gecko) "
          "Chrome/152.0.0.0 Safari/537.36 Edg/152.0.0.0"
      )
  }

response = requests.get(url, headers=headers, timeout=10)
response.raise_for_status()

# 학력 추출
soup = BeautifulSoup(response.text, "html.parser")

title = soup.select_one("#firstHeading")
print(title.get_text(strip=True) if title else "제목을 찾지 못했습니다.")
print(response.status_code)
```

- .find : 조건에 맞는 첫번째 요소
- .find_all : 조건에 맞는 모든 요소 - 리스트 반환
- .select_one : CSS 선택자로 첫번쨰 요소
- .select : CSS 선택자로 모든 요소 선택 → 리스트 반환
- .get_text : 요소 텍스트만
- .get("속성") :특정 속성값 추출
