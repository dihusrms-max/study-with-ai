### LLM 기반으로 태그

- LLM에게 문서를 그대로 던져주고 이것들을 주제가 뭐야? 질의
- 사라밍 붙일만한 이름을 예시로 같이 던져 주기

```python
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()
client = OpenAI()
```

```python
prompt = """
너는 단편 소설 작가야. 재미있게 대답을 잘 할 수 있어
너의 말투는 경상도 사투리를 써
간결하게 한국어로 3줄 이내로 써
"""
print(prompt)
```

```python
# 첫 요청
response = client.chat.completions.create(
    model="gpt-5.6-luna",
    messages=[
        {"role" : "user", "content": prompt},
        {"role" : "user", "content": "일어나 보니 좀비세상이 되어 버린 세상"}
        ]
)
```

```python
print(response.choices[0].message.content.strip())
```

```python
import pandas as pd
data = pd.read_csv("../data/11-1_뉴스정제.csv")
data.head(1)
```

```python
data['정제본문'][:10]
```

```python
#llm 에게 그냥 이름 지어달라고 할시 없는 내용이 생긴다
#문서에 없는 사실 추가 금지
#형식을 지키자
system_prompt = """
  너는 뉴스 문서의 핵심 주제를 파악하는 분류 전문가다.

  입력된 뉴스 대표 문서를 읽고, 문서의 중심 주제를 잘 나타내는
  간결한 한국어 토픽 이름을 만들어라.

  규칙:
  - 2~5어절로 작성한다.
  - 명사형 표현을 사용한다.
  - 문서에 실제로 드러난 핵심 내용만 반영한다.
  - 너무 넓거나 추상적인 표현은 피한다.
  - 설명, 번호, 따옴표, 마침표를 붙이지 않는다.
  - 토픽 이름만 한 줄로 출력한다.
  """
```

```python
documents = data["정제본문"].head(10)

topic_names = []

for document in documents:
    response = client.chat.completions.create(
        model="gpt-5.6-luna",
        messages=[
            {"role": "system", "content": sys_prompt},
            {
                "role": "user",
                "content": f"""
다음 뉴스 대표 문서의 토픽 이름을 만들어라.

[뉴스 대표 문서]
{document}
"""
            }
        ]
    )

```

```python
# 모델이 생성한 텍스트만 꺼내기
topic_name =
response.choices[0].message.content.strip()
topic_names.append(topic_name)


topic_names
```
