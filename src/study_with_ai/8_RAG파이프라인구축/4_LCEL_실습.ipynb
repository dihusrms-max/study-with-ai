### LCEL - 랭체인
- prompt : 프롬프트 템플릿으로 관리(파일관리)


- model: LLM 종류 들 전부 - openai, gemini, claude
                    로컬모델 : gemma, qwen
                    한국모델 : LG 엑사원, KT 마음, SK 에이닷, NV 바르코
- output
- parser : 답을 원하는 방식으로 다듬는 부분
- chain : chain = prompt | model | parser - 파이프라인 구성
- Langchain :각각 부품화 해서 조립하기 쉽게하는 라이브러리
- ->실행 : invoke
- 각각을 부품처럼 교체하기 용이함

```python
from Langchain_chat_models import init_Chat_model

model = init_Chat_model(

    model = "openai : gpt-5.6-luna"
    reasoning = {'effort':"none"}
)

# 가장 단순한 model 호출
result = model.invoke("파이썬에 대해서 알려줘")
result(result.text)
```
