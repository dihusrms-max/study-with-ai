```python
import sqlite3

conn = sqlite3.connect("library.db")
cur = conn.cursor()
```

```python
cur.execute(""" 
CREATE TABLE books(
    book_id INTEGER PRIMARY KEY,
    title TEXT,
    author TEXT,
    year   INTEGER

)
""")
```
