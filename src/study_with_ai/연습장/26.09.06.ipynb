```python
df - 데이터 프레임은 실제 데이터를 다루기 위한 도구
pd - 판다스는 데이터 처리를 위한 세부 도구 

데이터프레임 - 표전체
시리즈 - 열 하나 
컬럼은 속성과 같은 행
레코드는 속성 아래 값들 
인덱스는 보통 리스트에서 해당 행의 번호를 지정하여 조회 가능


보통 원본 - 정제본 - 집계본 으로 나눔 
정제를 하다보면 그룹으로 나누어 결측치 이상치등을 수정해야 하기 때문에 목적에 따라 여러파일이 만들어짐
집계본은 통계를 두어 중앙값, 평균 등 수학적 요소가 어떻게 들어가는 것이 좋을 지에 대해서 생각하면서 파일을 만듬
```

```python
isna는 not available으로 결측지 여부를 문자로 표시하여 존재여부 확인 가능
dropna으 경우 결측지를 제거해 버림
drop_duplicates()는 중복을 제거하여 행을 정리할 수 있음
gropby() 같은 기준값끼리 묶어서 정제본으로 만들 수 있음 
집게본을 만들 때 sum(), mean() 평균, median() 중앙값으로 편향 확인 가능
```
