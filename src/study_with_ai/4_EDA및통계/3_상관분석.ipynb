```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams["font.family"] = "Malgun Gothic"   # 한글 깨짐 방지(윈도우 기본 폰트)
plt.rcParams["axes.unicode_minus"] = False

df = pd.read_csv("../data/07-1_study_habits.csv") # r" " 사용으로 백슬래시 기능인 줄바꿈 무시 
df.head()
```

### 2. 산점도로 관계보기
- 두 변수 x, y에 놓고 점을 찍어서 비교
- 공부시간, 시험점수의 관계를 비교하는 것

```python
sns.scatterplot(data=df,
                x="공부시간" ,
                y="시험점수",
                hue="반")
plt.title("공부시간과 시험점수 관계")
plt.show()
```

### 3. 상관계수
- -1 에서 1사이의 값
- -1 에 근접할 수록 음의 상관관계
-  0 에 근접할 수록 관계가 없다

```python
#correlation - 상관관계

corr = df['공부시간'].corr(df['시험점수'])
corr
```

```python
### 4. 히트맵 - 여러 변수를 한 눈에 보기

corr_matrix = df[['공부시간', "시험점수", "용돈"]].corr() 
corr_matrix
```

```python
sns.heatmap(corr_matrix,
            annot=True,
            cmap='coolwarm',
            vmin=-1)
plt.title("히트맵")
plt.show()
```

- 공부시간과 시험점수는 상관계수가 0.78 -> 공부를 많이 할수록 점수가 높은 경향이 있다
- 용돈은 다른 변수와 거의 상관관계가 없다
- 유의해야 할 점 
    - 상관계수가 높다고 무조건 인과관계가 성립되지는 않는다
    - ex) 실제 원인을 볼 필요가 있음
- 상관계수가 0이다. 두 변수가 관계가 없다고 보는 것은 힘든 부분이 존재한다
