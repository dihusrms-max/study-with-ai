### 유의미한 차이인지 아닌지 
- 귀무가설 남녀의 전체 이용건수가 같다
- 대립가설 남녀의 이용 건수가 다르다

### p-Value
- p-value : 귀무가설이 맞다고 할 떄, 지금 본거 같은 차이가 우연히 나올때
- 유의수준 : (0.05) 이정도로 드물면 우연이 아니다라고 보겠다는 기준선

```python
from scipy import stats
# t 검정으로 p-value

t, p =  stats.ttest_ind(male, female, equal_var=False)
print(t,"t통계랑")
print(f"{p:4f}", "p-value")
```

- p-value 가 0.05 보다 작으면 -> 차이가 없다라는 귀무가설
- 대립가설
- 남녀 차이는 우연이 아니라 통계적으로 유의미하다


### 내판단이 틀릴 수 있따
- 1종 오류
    - 사실은 차이가 없는데 있따고 잘못 판단 (억울한 판정)
- 2종 오류
    - 사실은 차이가 있는데 없다고 놓친 판단(놓친 판정)
유의수준이 0.05라는 의미는 -> 1종 오류를 5%까지는 감수하겠다라는 약속

### t검정
- 가장 자주 찾는 질문 : 두 그룹의 평균이 다른가?
- 가장 기본적인 판단 기준

- 일표본 :한 그룹의 평균이 특정 기준과 다른가?
    - 예) 평균 이용 시간이 70분과 다른가?
- 독립표폰 : 서로 다른 두 그룹 비교
- 대응표본 : 같은 대상의 두 조건 비교
    - 같은 상권의 주중 vs 주말

### 1. 일표본 t검정 - 기준값과 다른가
- 따릉이 평균 이용 건수가 40건과 다른가?
- 0.05에 가까울 수록 다르다
- 0.05와 멀어 질수록 맞다

```python
t, p = stats.ttest_1samp(df_gender['이용건수'], 40)
print('평균이용건수', df_gender['이용건수'].mean())
print(p)
```

### 서로 다른 그룹 비교
- 남성과 여성의 평균 이용시간


```python
df_gender
df.head()
```

```python
df_gender['이용시간(분)/건별'] = (df_gender['이용시간(분)'] / df_gender['이용건수']).round(2)
df.head(3)
```

```python
male_time = df_gender[df_gender['성별']=='M']['이용시간(분)/건별']
male_time
```

```python
female_time = df_gender[df_gender['성별']=='F']['이용시간(분)/건별']
female_time
```

```python
t, p = stats.ttest_ind(male_time, female_time, equal_var=False)
print(f"남자평균 {male_time.mean():.2f}")
print(f"여자평균 {female_time.mean():.2f}")
print("p-value",p)
# (df_gender['이용시간(분)/건별'], 40)
# print('평균이용건수', df_gender['이용시간(분)/건별'].mean())
# print(p)
```
