```python
### 1. 그룹비교
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams["font.family"] = "Malgun Gothic"   # 한글 깨짐 방지(윈도우 기본 폰트)
plt.rcParams["axes.unicode_minus"] = False

df = pd.read_csv("../data/07-1_study_habits.csv") # r" " 사용으로 백슬래시 기능인 줄바꿈 무시 
df.head()
```

### 2. groupby을 사용한 그룹별 평균개수

```python
df.groupby("반")['시험점수'].mean().round(1) # round를 사용해서 반올림
```

```python
### 막대그래프 그리기

# sns.barplot(data=df, x='반',y='시험점수')
# plt.show()

plt.figure(figsize=(8, 5))

ax = sns.barplot(
    data=df,
    x="반",
    y="시험점수",
    estimator="mean",
    errorbar=None,
    color="#4C78A8"
)

ax.set_title("반별 평균 시험점수",
fontsize=16, pad=15)
ax.set_xlabel("반", fontsize=12)
ax.set_ylabel("평균 시험점수",
fontsize=12)

ax.set_ylim(0, 100)
ax.grid(axis="y", linestyle="--",
alpha=0.4)
ax.set_axisbelow(True)

# 막대 위에 평균값 표시
for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.1f",
        padding=3,
        fontsize=11
    )

sns.despine()
plt.tight_layout()
plt.show()
```

```python
### 4.boxplot으로 분포까지 비교
```

```python
sns.boxplot(data=df,
            x='반',
            y='시험점수')
plt.show()
```

```python
#### 5. hue를 사용해서 색깔로 구분해보기 hue를 사용해서 성별으로 구분이 가능하다
sns.boxplot(data=df,
            x='반',
            y='시험점수',
            hue="성별")
plt.show()
```
