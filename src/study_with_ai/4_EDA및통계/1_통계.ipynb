#### 범위, 분산, 표준편차
- 범위 : 최댓값 ~ 최솟값
- 분산, 표준편차 : 평균에서 얼마나 떨어져 있는 지 보는 통계값

```python
score = df['시험점수']
print("최댓값", score.max())
print("최솟값", score.min())
print("최빈값", score.mode())

score = df['시험점수'].describe()
score
```
