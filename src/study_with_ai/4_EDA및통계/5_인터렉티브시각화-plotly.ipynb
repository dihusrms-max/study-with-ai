```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go # 그래프 생성의 디테일 제어 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.rcParams["font.family"] = "Malgun Gothic"   # 한글 깨짐 방지(윈도우 기본 폰트)
plt.rcParams["axes.unicode_minus"] = False

df = pd.read_csv("../data/07-1_study_habits.csv") # r" " 사용으로 백슬래시 기능인 줄바꿈 무시 
df.head()
```

```python
# fig = px.scatter(data_frame=df,
#                  x='공부시간',
#                  y='시험점수',
#                  trendline= "ols",
#                  title='공부시간과 시험점수')
# fig.show()

fig = px.scatter(
    data_frame=df,
    x="공부시간",
    y="시험점수",
    trendline="ols",
    trendline_color_override="#E45756",
    title="공부시간과 시험점수의 관계",
    labels={
        "공부시간": "공부 시간 (시간)",
        "시험점수": "시험 점수"
    },
    template="plotly_white",
    hover_data={
        "공부시간": ":.1f",
        "시험점수": ":.1f"
    }
)

fig.update_traces(
    selector=dict(mode="markers"),
    marker=dict(
        size=11,
        color="#4C78A8",
        opacity=0.8,
        line=dict(
            width=1,
            color="white"
        )
    )
)

fig.update_layout(
    width=850,
    height=550,
    title=dict(
        text="공부시간과 시험점수의 관계",
        x=0.5,
        xanchor="center",
        font=dict(size=22)
    ),
    font=dict(
        family="Arial",
        size=13,
        color="#333333"
    ),
    plot_bgcolor="#F8FAFC",
    paper_bgcolor="white",
    xaxis=dict(
        showgrid=True,
        gridcolor="#D9E2EC",
        zeroline=False
    ),
    yaxis=dict(
        showgrid=True,
        gridcolor="#0D7D8100",
        zeroline=False
    ),
    legend=dict(
        bgcolor="rgba(255,255,255,0.7)"
    )
)

fig.show()
```

#### 다양한 차트 그리기

- 막대차트
- 원차트(pie) : 비율

```python
mean = df.groupby("반", as_index=False)['시험점수'].mean()
print(mean)
```

```python
mean_df = (
    df.groupby("반", as_index=False)["시험점수"]
    .mean()
    .rename(columns={"시험점수": "평균시험점수"})
)

fig = px.bar(
    data_frame=mean_df,
    x="반",
    y="평균시험점수",
    text_auto=".1f",
    title="반별 평균 시험점수",
    labels={
        "반": "반",
        "평균시험점수": "평균 시험점수"
    },
    template="plotly_white"
)
fig.update_layout(width=600, height=400)
fig.show()
```

```python
class_counts = df['반'].value_counts().reset_index()
class_counts
```

```python
fig = px.pie(data_frame=class_counts,
             names="반", 
             values="count",
             title="반별 인원 비율")
fig.update_layout(width=400, 
                  height=400, 
                  )  # 위쪽 여백)
fig.show()
```

```python
fig = px.pie(
      data_frame=class_counts,
      names="반",
      values="count",
      title="반별 인원 비율",
      hole=0.45,  # 가운데 빈 공간 비율
      color_discrete_sequence=["#4C78A8", "#F58518", "#54A24B", "#E45756"]
  )

# 조각 안에 반 이름과 퍼센트를 표시
fig.update_traces(
      textinfo="label+percent",
      textposition="inside",
      textfont_size=14,
      hovertemplate="%{label}<br>인원: %{value}명<br>비율: %{percent}<extra></extra>",
      pull=[0.03 if name != "C" else 0.15 for name in fig.data[0].labels]
)

fig.update_layout(
    width=500,
    height=500,
    title_x=0.5,  # 제목을 가운데 정렬
    margin=dict(t=70, b=70, l=30, r=30),
    legend=dict(orientation="h", y=-0.1, x=0.5, xanchor="center")
)

# 도넛 가운데에 전체 인원 표시
fig.add_annotation(
    text=f"전체<br>{class_counts['count'].sum()}명",
    x=0.5,
    y=0.5,
    showarrow=False,
    font=dict(size=18)
)

fig.show()
```

```python
fig = px.box(data_frame=df,
             x='반',
             y='시험점수',
             color='반'
             )
fig = px.histogram(

    data_frame=df,
    x='반',
    y='시험점수',

)

fig.update_layout(width=400, 
                  height=400, 
                  )  # 위쪽 여백)
fig.show()
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# 반 종류 확인
classes = sorted(df["반"].dropna().unique())

colors = [
    "#4C78A8",
    "#F58518",
    "#54A24B",
    "#E45756",
    "#B279A2"
]

fig = make_subplots(
    rows=1,
    cols=2,
    subplot_titles=(
        "반별 시험점수 분포",
        "시험점수 분포"
    )
)

for i, class_name in enumerate(classes):
    class_df = df[df["반"] == class_name]
    color = colors[i % len(colors)]

    # 왼쪽: Boxplot
    fig.add_trace(
        go.Box(
            x=[str(class_name)] * len(class_df),
            y=class_df["시험점수"],
            name=f"{class_name}반",
            marker_color=color,
            boxmean=True,
            showlegend=False
        ),
        row=1,
        col=1
    )

    # 오른쪽: Histogram
    fig.add_trace(
        go.Histogram(
            x=class_df["시험점수"],
            name=f"{class_name}반",
            marker_color=color,
            opacity=0.6,
            nbinsx=10
        ),
        row=1,
        col=2
    )

fig.update_layout(
    width=900,
    height=450,
    template="plotly_white",
    title="반별 시험점수 분석",
    title_x=0.5,
    barmode="overlay",
    legend_title="반"
)

fig.update_xaxes(
    title_text="반",
    row=1,
    col=1
)

fig.update_yaxes(
    title_text="시험점수",
    range=[0, 100],
    row=1,
    col=1
)

fig.update_xaxes(
    title_text="시험점수",
    range=[0, 100],
    row=1,
    col=2
)

fig.update_yaxes(
    title_text="학생 수",
    row=1,
    col=2
)

fig.show()
```

#### 여러 그래프를 한 화면에 출력
- 여러 그래프를 한 화면에 모으면 그게 대시보드




```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go

fig = make_subplots(
    rows=1,
    cols=2,
    subplot_titles=[
        "시험점수 분포",
        "공부시간 분포"
    ]
)

# 첫 번째 그래프: 시험점수 히스토그램
fig.add_trace(
    go.Histogram(
        x=df["시험점수"],
        name="시험점수",
        marker_color="#4C78A8",
        nbinsx=10
    ),
    row=1,
    col=1
)

# 두 번째 그래프: 공부시간 히스토그램
fig.add_trace(
    go.Histogram(
        x=df["공부시간"],
        name="공부시간",
        marker_color="#F58518",
        nbinsx=10
    ),
    row=1,
    col=2
)

fig.update_layout(
    width=900,
    height=450,
    title="시험점수와 공부시간 분포 비교",
    title_x=0.5,
    template="plotly_white",
    showlegend=False
)

fig.update_xaxes(title_text="시험점수",
row=1, col=1)
fig.update_yaxes(title_text="학생 수", row=1,
col=1)

fig.update_xaxes(title_text="공부시간",
row=1, col=2)
fig.update_yaxes(title_text="학생 수", row=1,
col=2)

fig.show()
```
