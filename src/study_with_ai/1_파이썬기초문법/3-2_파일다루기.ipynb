### 파일 다루기
- 우리가 다루는 데이터는 코드상에 없다
- 실제 데이터는 외부 별도 파일에 존재
- 외부에서 파일 불러오기 및 수정 방법 학습


- 지금까지 저장했던 리스트와 딕셔너리는 프로그램 종료 시 사라짐
1. csv 파일 : 엑셀 다루기 쉬운 형태
2. json 파일 : 웹에서 가져오는 데이터 형태
3. txt 파일 : 로그 데이터 등등 일반적 텍스트 형태


```python
# 파일 오픈
data = ["카리나","윈터","지젤","닝닝"]
file = open("test.txt","w",encoding="UTF-8")
for i in data:
    file.write(i+"\n")
file.close()
```

```python
# 파일 오픈 2
# 파일 오픈
data = ["카리나","윈터","지젤","닝닝","빈하늘"]
with open("test.txt","w",encoding="UTF-8") as file:
    for i in data:
        file.write(i+"\n")
    file.close()
```

```python

# 파일 읽기
# 1. 모두 읽기 - read
# 2. 한줄 읽기 - readline - 용량 클 때 사용 
# 3. 한줄을 리스트로 읽기 - readlines

with open("test.txt","r",encoding="utf-8") as file:
    text = file.read() # 전체
    print(text)
```

```python
with open("test.txt","r",encoding="utf-8") as file:
    while True:
        text = file.readline().strip() # 1줄만

        if text=="": # 더 이상 없음 None
            break # 종료
        #종료 조건 필요
        print(text)
```

```python
with open("test.txt","r",encoding="utf-8") as file:
    text = file.readlines() # 전체를 리스트로 
    print(text)
```

```python
# 파일 내용 추가 모드는 a
with open("test.txt","a",encoding="utf-8") as file:
    file.write("민서율") # 1줄만
   
```
