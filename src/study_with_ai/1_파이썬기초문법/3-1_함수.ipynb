#### args - 매개변수의 갯수가 정해지지 않았을 때
  - 필수 항목 → 일반 매개변수
  - 선택 항목(기본값 있음) → 기본값 매개변수
  - 여러 값, 개수 미정 → *args
  - 여러 이름표 값, 항목이 매번 다름 → **kwargs

```python
def introduce_1(name, age, city,*hobbies,**extra_info):
    print(f"이름은 {name}이며 나이는 {age}이고 {city}에 거주하며 추가로 {hobbies}를 하곤 하며 {extra_info}도 합니다")

def introduce(name, age, city, *hobbies, **extra_info):
      info = ", ".join(
          f"{key}: {value}"
          for key, value in extra_info.items()
      )

      print(f"이름: {name}, 나이: {age}, 거주지: {city}, 추가 정보: {info}")

    # *args --> 튜플, *hobbies로 직관적으로 수정 가능 - join()사용 권장 투플 () 형태 그대로 출력 되는 사항 존재
    # **kwargs --> 딕셔너리 - 키:값으로 입력되어야 한다
```

```python
introduce("김민수",22,"서울","스포츠", 취미="테니스")
```

```python
def introduce(name, age,**kwargs):
    print("name?", name)
    print("age?", age)

    for key, Value in kwargs.items():
        print(key, ":", Value)

# .items() - ()는 실행하라는 의미, 안에는 인자값이 들어가 변수로 전달 가능
# 그런데 인자를 채우는 지 여부를 모를 경우가 있다 
# 이때 help() 사용 - help(kwargs.items)로 결과를 출력하자 
```

```python
introduce("karina",27,취미="쇼핑", 거주지="서울",번호="010-4524-3235")
```
