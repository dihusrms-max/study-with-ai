```python
month_bill = 300_000
lunch_bill = 8_500 
cafe_bill = 4_500
book_bill = 24_000
교통비 = 65_000

total_expenses =  (lunch_bill*12) + (cafe_bill*6) + book_bill + 교통비
print(f"이번 달 전체 지출 : {total_expenses}")

remaining_bill = month_bill - total_expenses
print(f"남은용돈 : {remaining_bill}")
remaining_percent = (remaining_bill / month_bill) * 100
print(f"남은용돈_비율 : {remaining_percent:.2f}%") # 천단위 구분자 표시
round(remaining_percent, 2)
```

```python
starting_balance = 500_000 #시작 잔액

part_time_income = 320_000 #아르바이트 수입
allowance_income = 150_000 #용돈

food_expense = 185_000 #식비
transport_expense = 62_000 #교통비
phone_expense = 45_000 #통화비
hobby_expense = 78_000 #활동비
savings = 120_000 #저축

total_income = part_time_income + allowance_income
total_expenses = food_expense + transport_expense + phone_expense + hobby_expense + savings
final_balance = starting_balance + total_income - total_expenses
print(f"최종 잔액 : {final_balance}")
round(final_balance, 2)
```
