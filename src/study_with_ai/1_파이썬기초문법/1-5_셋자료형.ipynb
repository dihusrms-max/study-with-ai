### 셋자료형
- 중복 제거용
- {} 중괄호 사용, 딕셔너리와 구분 필요!!


```python
# 합집합 : 중복 제거 전제 값 --> | 또는 union 을 사용
data1 = {1, 2, 3}
data2 = {3, 4, 5}

data1|data2
```

```python
data3 = {1, 2, 3,3,3,5}
data4 = set(data3)
data4
```

```python
print(data1)
print(data2)
```

```python
print(data1|data2)
print(data1.union(data2))
```

```python
#### 교집합 - 공통된 겹치는 부분 

data1 & data2
```

```python
data1.intersection(data2)
```

```python
#차집합 - 한쪽에만 있는 값
data1 - data2
```

```python
data1.difference(data2) # 앞 부부분 뒷부분 위치 변경시 결과가 달라짐
```

```python
data2.difference(data1)
```
