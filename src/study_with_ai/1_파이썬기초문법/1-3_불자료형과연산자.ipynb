### 연산자
- 산술 연산자
- 비교 연산자 : ==, !=,>, <, >=, <= 같다, 다르다, 크다, 작다, 크거나 같다, 작거나 같다
- 논리 연산자 : and, or, not 그리고 또는 아닌 : 여러 조건을 묶을 때




```python
# 같다
3 == 4
```

```python
3 == 3 # True
```

```python
3 =="3" # False
```

```python
3 == int("3") # True
```

```python
"홍길동" == "홍길동" # True
```

```python
# 연습 시험점수가 60점 이상인지 확인
score = 75
passed = (score >= 60)
print(passed) # True
```

```python
"홍길동" != "이순신" # True
```

### 논리 연산자
- and : 둘다 참 일때 
- or : 둘 중 하나가 참일 때 
- not : 둘 중 전부 거짓일 때

```python
age = 28
print(20 <= age <=30)
```

```python
age = 8
print(not(age == 8))
```

```python
my_id = 'song'
my_id != ""
```
