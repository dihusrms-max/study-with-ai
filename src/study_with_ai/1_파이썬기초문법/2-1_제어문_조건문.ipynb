### 조건문

- if 
- if ~ else

#### 조건문 중요 사항 
- 간단하게 코드는 위에서 아래 방향으로 실행
- 특성 상황에 따라서 실행여부를 결정할 수 있게 조건문이 필요
- ex) 
    - 나이 19살 - 성인, 아니다 - 미성년자
    - 점수가 90이상 - A, 80점 이상 - B
    - 물건 재고 있을 경우 -> 주문 가능, 재고 없는 경우 -> 품절 
- 다양한 분기점 에서 실행 여부를 정해주는 if 조건문

#### if ~elif ~ else

```python
score = 82

if score > 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"
print(f"학점은 {grade}입니다.")
```

```python
temperature = 30

if temperature >= 28:
    print("반팔 추천드립니다")
elif temperature >= 15:
    print("긴팔 추천드립니다")
else:
    print("패딩 추천드립니다")
```

```python
price = 100000
if price >= 300_000:
    print(f"10% 할인으로 {price * 0.1}원입니다")
elif price >= 200_000:
    print(f"5% 할인으로 {price * 0.05}원입니다")
elif price >= 100_000:
    print(f"5000원 할인으로{price - 5000}원 입니다")
else:
    print(f"할인 없음{price}원")
```

```python
price = 100000

if price >= 300_000:
    discount = price * 0.1
    message = "30만원 이상 구매시 10% 할인"
elif price >= 200_000:
    discount = price * 0.05
    message = "20만원 이상 구매시 5% 할인"  
elif price >= 100_000:
    discount = 5000
    message = "10만원 이상 구매시 5000원 할인"
else:
    discount = 0
    message = "할인 없음"

print(message)
print(f"할인 금액은{discount}이고 최종{price - discount}원입니다.")
```

#### 리스트 컴프리헨션 
- for문으로 정리한 데이터가 필요할 때 리스트 사용시
- 짧은 한줄로 표현하고 싶을 때

```python
data = [] # 1. 데이터 생성

for i in range(1,6):
     print(i**2) # 데이터 출력 이상 여부 확인
     data.append(i**2) #3. 확인 후 결과를 빈 리스트안에 담기
data
```
