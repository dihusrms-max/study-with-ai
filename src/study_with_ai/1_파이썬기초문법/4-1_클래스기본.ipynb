### 클래스
- 공통된 속성과 한번만 정의 이를 기반으로 객체를 효율적 관리를 위한 개념
- 설계도 개념


```python
# 간단한 클래스 작성

class Human:
    name = ""
    age = 0
```

```python
hyeon = Human()
hyeon
```

```python
hyeon.name = "카리나"
hyeon.age = "27"
```

```python
hyeon.name
```

```python
hyeon.age
```

```python
# 메서드가 있는 클래스 생성 - 메서드는 클래스에서 사용하는 함수

class Human2:
    # 속성 정의
    name = ""
    age = 0

    # 메서드 정의

    def introduce(self): # 현재 클래스 속성을 가리켜서 사용
        print(f"{self.name} 입니다")
        print(f"{self.age} 입니다")

```

```python
yth = Human2()
yth.name = "윤택한"
yth.age = 35
```

```python
yth.__dict__
```

```python
yth.introduce()
```

```python
# 클랜스 생성자  - 초기 입력
class book:
    #초기에 반드시 잇어야 하는 항목이 있을 경우
    def __init__(self,title,author):
        self.title = title
        self.author = author

        #메서드 추가 - 간단한 책소개 기능 추가

    def show_info(self):
        print(f"""
이 책의 제목은 {self.title} 이고
저자는 {self.author}입니다.
많이 사랑해주세요        
""")
```

```python
my_book = book("고수","카리나,윈터,닝닝,지젤")
```

```python
my_book.show_info()
```
