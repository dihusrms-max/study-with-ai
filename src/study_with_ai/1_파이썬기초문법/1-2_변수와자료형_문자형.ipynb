```python
# 문자열 포맷팅 문자열 틀을 고정하고 바뀔 부분만 바꿀때 포맷팅 - f-strings

name = '송하예'
year = 2026
f_greeting = f"{name}님 안녕하세요! {year}년입니다."
print(f_greeting)

#format() 사용
f_greeting = "{name}님 안녕하세요! {year}년입니다.".format(name=name, year=year)
print(f_greeting)

#%s %d 사용
f_greeting = "%s님 안녕하세요! %d년입니다." % (name, year)
print(f_greeting)
```

```python
#문자열 메서드
#- 메서드 : .을 사용
#- 원본은 바뀌지 않고 새 문자열을 반환
#- 문자열에서 사용하는 다양한 도구 존재

text = "Hello, World!"
print(text.upper())  # HELLO, WORLD!
print(text.lower())  # hello, world!
print(text.strip())  # Hello, World! (공백 제거)
print(text.replace("World", "Python"))  # Hello, Python!
```

```python
raw = " Hello, World! "
print(raw.strip())  # Hello, World! (공백 제거)
print(raw.lstrip())  # Hello, World! (왼쪽 공백 제거)
print(raw.rstrip())  # Hello, World! (오른쪽 공백 제거)
```

```python
#문자열 나누기

text = "2026-08-31"
text_blank = "2026 08 31"
text_slash = "2026/08/31"

print("원본", text)
print("하이픈으로 나누기", text.split("-"))
print("공백으로 나누기", text_blank.split())
print("슬래시로 나누기", text_slash.split("/"))
```

```python
student_info = "홍길동/20/서울"
name, age, city = student_info.split("/")
print(f"이름: {name}, 나이: {age}, 도시: {city}")
```

```python
#문자열 변경
student_info = "홍길동/20/서울"
name, age, city = student_info.replace("/", " ").split(" ")
print(student_info)
```

```python
date = " 2026.08.31 "

 # 1. 점 .을 기준으로 나누기
 # 2. 연도, 월, 일을 각각 year, month, day에 저장하기
 # 3. 아래 형식으로 출력하기

date = date.split(".")
year, month, day = date[0].strip(), date[1].strip(), date[2].strip()
print(f" 연도:{year}\n 월:{month}\n 일:{day}")
```
