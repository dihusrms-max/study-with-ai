```python
#점수에서 영어→95 수정, 과학(80) 추가, 총점·평균·최고점을 출력하세요. (max(...values()))
점수 = {"국어": 90, "영어": 85, "수학": 100}

# TODO
점수["영어"] = 95
점수["과학"] = 80
# 과학 추가:
총점 = 점수["국어"] + 점수["영어"] + 점수["수학"]
평균 = 점수["국어"] + 점수["영어"] + 점수["수학"] / len(점수)
최고 = max(점수.values())
print("총점:", 총점, "/ 평균:", 평균, "/ 최고:", 최고)
# 예상: 총점 365 / 평균 91.25 / 최고 100
```
