## 문제 3. 메서드 추가 · ⭐⭐ 기본

`학생` 클래스에 `소개` 메서드를 넣어 `"OO입니다. 25살이에요."`를 출력하게 하고, 호출하세요.

```python
# TODO: class 학생 안에 def 소개(self): print(...)

class 학생 : 
    def __init__(self,이름,나이):
        self.이름 = 이름
        self.나이 = 나이

    def introduce(self): #매개변수에는 이름, 나이를 2번 사용할 필요가 없다
        print(f"{self.이름}입니다. {self.나이}살이에요") # self 객체에 접근하여 이름을 가져올 수 잇게 구성

```

```python
학생1 = 학생("김민수",24)
학생1.introduce()
```

## 도전. 책 클래스 · ⭐⭐⭐ 도전

`제목`·`저자`를 받는 클래스 `책`을 만들고, `정보` 메서드로 `"제목 - 저자"`를 출력하게 하세요.

```python
# TODO: class 책: __init__(self, 제목, 저자) / 정보 메서드


class 책:
    def __init__(self,제목,저자):
        self.제목 = 제목
        self.저자 = 저자
        
    def 정보(self):
        return print(f"{self.제목} - {self.저자}")
        
```

```python
책1 = 책("사람","카리나")
책1.정보()
```
