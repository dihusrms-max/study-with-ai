```python
#숫자로 된 리스트를 쉽게 만드는 방법을 배워보기
nums = list(range(1,5,1)) # 처음 중간 증가량 
print(nums)
nums = list(range(1,5,2))
print(nums)
```

```python
# remove 지우기 - 일치된 값 앞에서 1개 
nums = [10, 20, 30, 50,30, 40, 50, 60, 70]
nums.remove(30)
nums
```

```python
# pop 지우기 - 뒤에 값 -> 지운 값 기억
nums = [10, 20, 40, 50, 60, 70]
nums.pop(2)
nums=nums.pop()
```

### 딕셔너리
- 딕셔너리는 키와 밸류로 구성
- 딕셔너리 사용 이유
    key : value 연결되어 있어 key 사용 시 value를 가져 올 수 있음
    ex) 리스트 순서 번호 찾기 - 점수[0],점수[1] - 점수 조회 - 인덱스 번호 조회 시 - 여러 해당 인덱스 다수 존재하여 찾는 데 한계
- 이름표가 붙은 값
-{} key:value 나타냄
- 딕셔너리[key] 그 값 가져올 수 있음
- 딕셔너리 안에 리스트 튜플 사용가능 




```python
ai_family_dic = {"강기현":"01",
                 "송수림":"78",
                 "이현근":"00",
                 "김병주":["붉은사막","팰월드"],
                 "김수연":{"취미":"조승우덕질","출생년도":"00"}}
```

```python
ai_family_dic["강기현"]
```

```python
ai_family_dic["김수연"]
```

```python
ai_family_dic["유종혁"]={"취미":["게임","보드게임"],"출생년도":"89"}
ai_family_dic
```

```python
ai_family_dic["백성현"]={"취미":["유튜브 음악"],"출생년도":"79"}
ai_family_dic["주승우"]={"취미":["음악감상"],"출생년도":"01"}
ai_family_dic
```
