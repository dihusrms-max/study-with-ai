```python
#숫자로 된 리스트를 쉽게 만드는 방법을 배워보기
nums = list(range(1,5,1)) # 처음 중간 증가량 
print(nums)
nums = list(range(1,5,2))
print(nums)
```

```python
# 리스트 수정
nums = [1,2,3,4,5]
nums[2] = 20
nums 
```

```python
nums = [1,2,3,4,5]
nums[1:2] = [30,40]
nums 
```

```python
### 리스트 삭제

nums = list(range(10,51,10))
nums
```

```python
# del 요소 지우기 인덱싱

del nums[3]
nums
```

```python
# remove 지우기 - 일치된 값 앞에서 1개 
nums = [10, 20, 30, 50,30, 40, 50, 60, 70]
nums.remove(30)
nums
```

```python
# pop 지우기 - 뒤에 값 -> 지운 값 기억
nums = [10, 20, 40, 50, 60, 70]
nums.pop(2)
nums=nums.pop()
```

```python
# 리버스
nums = [15,10,7,30,15]
#nums.sort()
nums.reverse()
nums
```
