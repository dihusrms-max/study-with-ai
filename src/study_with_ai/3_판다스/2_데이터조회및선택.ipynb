### 4. 행 선택 iloc 
- 몇 번째인지 순서를 꺼낸다
 

```python
df.iloc[3:5,2:4]
```

```python
df.iloc[0:4]
```

```python
df.loc[0:3][["반","성별","국어","영어"]]
df.loc[0]
```

```python
df[df['성별']=="남"]
df[df['국어']>=80]
# df[df['수학']>=80] # 결적으로 전처리 필요
```

#### 5. 여러 조건 - &그리고, | 또는

- 조건이 여러개면 and , or를 사용
- 

```python
df[(df['반']=="A") & (df['국어'] >= 85)]
# 연습 
```

```python
df
```

```python
# 성별이 여자이면서 수학 점수가 70점 이상인 학생
df[(df['성별']=="여") & (df['영어'] >= 90)]
```

```python
# 성별이 여자이면서 국어 점수가 80점 이상인 학생
df[(df['성별']=="여") & (df['국어']>=80)]
```

```python
df[(df['국어']>=86) | (df['영어']>=90)]
```
