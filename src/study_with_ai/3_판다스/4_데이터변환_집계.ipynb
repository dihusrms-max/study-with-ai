```python
columns = ['공원명','공원구분','공원면적']

clean_df = df[columns]
clean_df
clean_df['큰공원여부'] = clean_df['공원면적'] >= 50_000
clean_df.head()
```

### 2. 면적을 기준으로 보고 싶다 -> 정렬 개념
- 큰값부터 정렬
- 작은값부터 정렬

```python
clean_df.sort_values("공원면적", ascending=False)
```

##### 3. 특정 그룹별로 한번 집계(통계게산)
- groupby
- 공원 종류별 묶어서 개수, 합계, 평균 면적 계산
- 어떤 유형의 공원이 어떤 도시에 많이 있는 지 

```python
# 공원면적이 0인 값, 호근 얼마 이하인 값들은 제거 예정
clean_df.groupby("공원구분")["공원면적"].count()
summary=clean_df.groupby("공원구분")["공원면적"].agg(['count','sum',"mean",'median'])
summary.sort_values("count", ascending=False)
```

  - 일반 생활공원: 근린공원, 소공원, 어린이공원, 가로공원, 마을마당
  - 자연형 공원: 수변공원, 도시농업공원
  - 특수목적 공원: 문화공원, 역사공원, 주제공원, 체육공원, 묘지공원
  - 확인 필요: 기타, 기타공원

##### 인사이트
- 어린이공원, 소공원은 개수가 많지만 그키가 작다
- 근린공원은 평균면적이 크다 
- 개수 1위는 어린이공원이지만, 전체면적으로 보면 주 공원은 근린공원이다.



### 4. 교차표 만들기 - pivot_table
- 어떤 두 칼럼(범주) 를 가로, 세로로 교차시켜 요약
- 관리기반 X 공원구분의 평균 면적을 같이 보려고 할 때






```python
table_df = clean_df.pivot_table(index="공원구분",
                          columns="큰공원여부",
                          values="공원명",
                          aggfunc="count",
                          fill_value=0)
table_df

```

```python
table_df = df.pivot_table(index="공원구분",
                          columns="큰공원여부",
                          values="공원명",
                          aggfunc="count",
                          fill_value=0)
table_df
```

```python
melt_df = table_df.reset_index()
melt_df.columns
```

```python
melt_df = table_df.reset_index().melt(id_vars="공원구분",
                          var_name="큰공원여부",
                            value_name="공원수",
                          )
melt_df
```

```python
clean_df['규모'] = clean_df['공원면적'].apply(lambda X: "큰공원" if X >=30_000 else "작은공원")
# 세분화 해본다면
# 50000이상 -> 큰공원
# 30000 ~ 50000 -> 중간공원
# 30000이하 -> 작은 공원

clean_df['규모'] = clean_df['공원면적'].apply(lambda x: "큰공원" if x >=50000 else ("중간공원" if x>=30000 else "작은공원"))
clean_df.head()
```

```python
def cleaned_fuc(x):
    if x >= 50_000:
        return "큰공원"
    elif x>= 30_000:
        return "중간공원"
    else:
        "작은공원"


clean_df['규모'] = clean_df["공원면적"].apply(cleaned_fuc)
clean_df.head()
```

```python
clean_df['규모cut'] = pd.cut(clean_df['공원면적'],
                           bins=[0,30_000,50_000,float('inf')],
                           labels=['큰공원','중간공원','작은공원'])
```
