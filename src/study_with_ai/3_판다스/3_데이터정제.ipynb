### 6. 타입 수정
- 숫자가 아닌 부분은 -> None
- pd.to_numeric() 

```python
df['수학']=pd.to_numeric(df['수학'], errors="coerce") # 숫자가 아닌 부분은 -> None
df
```

```python
df['수학']=df['수학'].fillna(df['수학'].mean())
df # 숫자가 아닌 부분은 -> None
```
