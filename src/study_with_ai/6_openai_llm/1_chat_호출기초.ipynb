### LLM 대화 기본 구조
- messages : 대화 목록
    - ai 대화
    - 내 대화
    - ai 대화
    - 내 대화
    - 맥락을 보존하려면 이 대화 내용 전체가 들어간다
    - 나중에는 최적화
- role : 역할
    - system : AI의 성격 지정, 기장 원초적인 지시, 규칙 담김 맨앞에 한번
    - user(human) : 사람이 하는 말
    - assistatant(ai) : AI가 답변한 말

```python
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv() 
client = OpenAI()
```

```python
response = client.chat.completions.create(

    model= "gpt-5.6-luna"
    messages=[{

        "role" : "user",
        "content" : " 파이썬의 장단점 추천해줘 3줄 내외로"
    }]
)

print(response)
```
