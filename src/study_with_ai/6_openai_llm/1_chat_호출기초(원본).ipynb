### LLM 대화 기본 구조
- messages : 대화 목록
    - ai 대화
    - 내 대화
    - ai 대화
    - 내 대화
    - 맥락을 보존하려면 이 대화 내용 전체가 들어간다
    - 나중에는 최적화
- role : 역할
    - system : AI의 성격 지정, 기장 원초적인 지시, 규칙 담김 맨앞에 한번
    - user(human) : 사람이 하는 말
    - assistatant(ai) : AI가 답변한 말

```python
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv() 
client = OpenAI()
```

```python
response = client.chat.completions.create(

    model= "gpt-5.6-luna",
    messages=[{

        "role" : "user",
        "content" : " 파이썬의 장단점 추천해줘 3줄 내외로"
    }]
)

print(response)
```

```python
print(response.choices[0].message.content)
```

```python
print(response.choices[0].finish_reason)
```

```python
print(response.usage.total_tokens)
```

role 나눠서 써보기
- system 프롬프트를 적용하기 위한 성격

```python
# 대화
# 시스템프롬프트
# user 프롬프트
# AI 답변 -> response


response = client.chat.completions.create(

    model= "gpt-5.6-luna",
    messages=[{

        "role" : "system",
        "content" : " 너는 초등학생에게 설명을 아주 쉽게 잘하는 선생이자 추리를 좋아해, 쉬운말로 써"
        """ 말투는 경상도 사투리 """
    },
    {
        "role" : "user",
        "content" : " 대한민국 20대 30대 취업시장 5줄 내외로 브리핑해줘"

    }
    ]
)

print(response.choices[0].message.content)
```

# 대화 이어나가기
- 지난 대화는 기억하지 못한다
- 지난 대화는 리스트에 담아 놓기


```python
messages = [
    {"role":"user", "content":"너는 하츄핑이니?"},
    {"role":"assistant", "content":"그래 나는 하츄핑이야"},
    {"role":"user", "content":"하츄핑아 노래 하나 불러줘"},
]

response = client.chat.completions.create(
    model="gpt-5.6-luna",
    messages=messages
)
print(response.choices[0].message.content)
```

```python
messages.append({'role': 'assistant', 'content': response.choices[0].message.content})
messages
```

#### 스트리밍

- 답이 다 만들어진 후 받지 않고 조각조각으로 만들어질때마다 받고 할 예정

```python
stream = client.chat.completions.create(
    model="gpt-5.6-luna",
    messages=[{
        'role': "user", "content": "AI 활용방법을 배우면 좋은 점은?"
    }
    ],
    stream=True
)
pieces = []

for chunk in stream:
    piece = chunk.choices[0].delta.content
    # print(pieces, end="", flush=True)
    pieces.append(piece)
```
