### LLM 대화 기본 구조
- messages : 대화 목록
    - ai 대화
    - 내 대화
    - ai 대화
    - 내 대화
    - 맥락을 보존하려면 이 대화 내용 전체가 들어간다
    - 나중에는 최적화
- role : 역할
    - system : AI의 성격 지정, 기장 원초적인 지시, 규칙 담김 맨앞에 한번
    - user(human) : 사람이 하는 말
    - assistatant(ai) : AI가 답변한 말

```python
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv() 
client = OpenAI()
```

```python
response = client.chat.completions.create(

    model= "gpt-5.6-luna",
    messages=[{

        "role" : "user",
        "content" : " 파이썬의 장단점 추천해줘 3줄 내외로"
    }]
)

print(response)
```

```python
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv() 
client = OpenAI()
```

```python
stream = client.chat.completions.create(
    model="gpt-5.6-luna",
    messages=[
        {
            "role": "system",
            "content": (
                "법률 상담을 제공하되 변호사를 대신하지 않는다는 점을 안내하고, "
                "사실관계와 서류 내용을 먼저 확인한뒤 신중하게 답변하세요."
            ),
        },
        {
            "role": "user",
            "content": "이 사건에 대해서 법률상담해줘",
        },
        {
            "role": "assistant",
            "content": "고객님 어떤 사건에 대해서상담이 필요하신가요?",
        },
        {
            "role": "user",
            "content": "오늘 사건번호로 고소장이 집으로 날아왔어",
        },
        {
            "role": "assistant",
            "content": "어떤 유형의 고소장인지 알려주시겠어요?",
        },
        {
            "role": "user",
            "content": "잘 모르겠어. 왜 이런 게 날아왔는지 모르겠어.",
        },
        {
            "role": "assistant",
            "content": "혹시 짐작 가는 상황이나 기억나는 일이 있을까요?",
        },
    ],
    stream=True,
)

pieces = []

for chunk in stream:
    piece = chunk.choices[0].delta.content or ""

    print(piece, end="", flush=True)
    pieces.append(piece)

full_response = "".join(pieces)
```

### 법률상담: 긴급 여부에 따른 답변 분기
출석기한과 경찰의 현장 조치 여부를 먼저 확인한 뒤, 긴급하면 짧게 안내한다.

```python
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()
client = OpenAI()
```

```python
# 긴급 여부
deadline = input("출석기한 임박? (1: 예, 2: 아니오, 3: 모름): ")
police = input("경찰의 휴대폰 수색·체포 언급? (1:예, 2: 아니오, 3: 모름): ")

urgent = deadline in {"1", "3"} or police in {"1","3"}

if urgent:
    style = "긴급 상황이므로 해야 할 일만 3~4줄로답하고 냥냥체를 써줘."
else:
    detail = input("답변 방식? (1: 간단히, 2: 자세히): ")
    style = "핵심만 짧게 답하고 냥냥체를 써줘" if detail =="1" else "절차와 이유를 자세히 답하세요."


system_prompt = f"""
법률정보 안내 AI입니다. 변호사를 대신하지 않으며
법률 결과를 단정하지 마세요.
사실관계를 먼저 확인하고 신중하게 답변하세요.
체포나 휴대폰 제출 요구가 있으면 물리적으로 저항하
지 말고
체포 사유, 영장 여부, 임의제출 여부를 확인한 뒤 변
호사에게 연락하도록 안내하세요.

금지: 거짓말, 증거 조작·삭제·은닉, 도주, 물리적 저
항을 안내하지 마세요.
확인되지 않은 법률·판례를 만들지 마세요.
{style}
"""

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": "고소장을 받았고경찰 대응이 걱정됩니다."},
]


def ask_ai():
    stream = client.chat.completions.create(
        model="gpt-5.6-luna",
        messages=messages,
        stream=True,
    )

    answer = ""

    for chunk in stream:
        piece = chunk.choices[0].delta.content or""
        print(piece, end="", flush=True)
        answer += piece

    return answer


while True:
    question = input("\n질문 입력 (종료: exit):").strip()

    if question.lower() == "exit":
        break

    messages.append({"role": "user", "content":question})
    answer = ask_ai()
    messages.append({"role": "assistant", "content": answer})
```
