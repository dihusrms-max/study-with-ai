```python
from pathlib import Path
import gzip
import json
import zipfile

ZIP_PATH = Path("C:/study-contests/data/open.zip")
target_id = "PPS-DEV-01"

with zipfile.ZipFile(ZIP_PATH) as z:
    with gzip.GzipFile(fileobj=z.open("dev.jsonl.gz")) as f:
        for line in f:
            item = json.loads(line)
            if item["id"] == target_id:
                record = item
                break

text = record["docs"][0]["text"]
path = Path("C:/study-contests/outputs/PPS-DEV-01_공고문.txt")
path.write_text(text, encoding="utf-8")

print("저장 완료:", path)
```
